```python
import numpy as np
import pandas as pd
```

```python
df= pd.read_csv("bank_data_raw.csv")
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
print(df["CustomerID"].duplicated().sum())
print(df["CustomerID"].describe())
```

```python
dupe_ids=df[df["CustomerID"].duplicated(keep=False)].sort_values("CustomerID")
dupe_ids.head(20)
```

```python
# Sort so that rows with a non-null Age come first within each CustomerID group,
# then keep only the first occurrence of each CustomerID
df = df.sort_values("Age", na_position="last")
df = df.drop_duplicates(subset="CustomerID", keep="first")
df = df.reset_index(drop=True)

print("Remaining duplicate CustomerIDs:", df["CustomerID"].duplicated().sum())
print("New row count:", len(df))
```

```python
df["Customer Name"].apply(lambda x: x != x.strip()).sum()
```

```python
df["Customer Name"]=df["Customer Name"].str.strip()
```

```python
df["Customer Name"].apply(lambda x: x != x.strip()).sum()
```

```python
print("Missing:", df["Age"].isnull().sum())
print(df["Age"].describe())
print("Impossible ages (<18 or >100):")
print(df[(df["Age"] < 18) | (df["Age"] > 100)]["Age"].value_counts())
```

```python
df.loc[(df["Age"] < 18) | (df["Age"] > 100), "Age"] = np.nan
```

```python

print("Total missing Age after flagging outliers:", df["Age"].isnull().sum())
```

```python
median_age=df["Age"].median()
df["Age"]=df["Age"].fillna(median_age)
```

```python
print("Median used:", median_age)
print("Remaining missing:", df["Age"].isnull().sum())
print(df["Age"].describe())
```

```python
print("Missing:", df["CreditScore"].isnull().sum())
print(df["CreditScore"].describe())
print("Impossible scores (<300 or >850):")
print(df[(df["CreditScore"] < 300) | (df["CreditScore"] > 850)]["CreditScore"].value_counts())
```

```python
df.loc[(df["CreditScore"] < 300) | (df["CreditScore"] > 850), "CreditScore"] = np.nan
```

```python
print("Total missing CreditScore after flagging outliers:", df["CreditScore"].isnull().sum())
```

```python
median_credit = df["CreditScore"].median()
df["CreditScore"] = df["CreditScore"].fillna(median_credit)

print("Median used:", median_credit)
print("Remaining missing:", df["CreditScore"].isnull().sum())
print(df["CreditScore"].describe())
```

```python
df["Country"].value_counts()
```

```python
# Step 1: normalize casing (lowercase, then capitalize properly)
df["Country"] = df["Country"].str.strip().str.lower()

# Step 2: fix the one true synonym
df["Country"] = df["Country"].replace({"deutschland": "germany"})

# Step 3: capitalize nicely (germany -> Germany)
df["Country"] = df["Country"].str.capitalize()

print(df["Country"].value_counts())
```

```python
df["Gender"].value_counts()
```

```python
df["Gender"] = df["Gender"].str.strip().str.lower()
df["Gender"] = df["Gender"].replace({"m": "male", "f": "female"})
df["Gender"] = df["Gender"].str.capitalize()

print(df["Gender"].value_counts())
```

```python
print("Missing:", df["Tenure"].isnull().sum())
print(df["Tenure"].describe())
print(df["Tenure"].unique())
```

```python
print(df["Balance"].dtype)
df["Balance"].head(20)
```

```python
# Step 1 & 2: strip symbols and convert to numeric
df["Balance"] = (
    df["Balance"]
    .astype(str)
    .str.replace("€", "", regex=False)
    .str.replace("EUR", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
    .astype(float)
)

print(df["Balance"].dtype)
print(df["Balance"].describe())

# Step 3: check for negative balances
print("Negative balances:", (df["Balance"] < 0).sum())
```

```python
df.loc[df["Balance"]<0,"Balance"].isnull().sum()
```

```python
df.loc[df["Balance"] < 0, "Balance"] = np.nan
print("Missing Balance after flagging negatives:", df["Balance"].isnull().sum())

median_balance = df["Balance"].median()
df["Balance"] = df["Balance"].fillna(median_balance)

print("Median used:", median_balance)
print("Remaining missing:", df["Balance"].isnull().sum())
print(df["Balance"].describe())
```

```python
print("Missing:", df["NumOfProducts"].isnull().sum())
print(df["NumOfProducts"].describe())
print(df["NumOfProducts"].unique())
```

```python
df["HasCrCard"].value_counts()
```

```python
df["HasCrCard"] = df["HasCrCard"].astype(str).str.strip().str.lower()

df["HasCrCard"] = df["HasCrCard"].replace({
    "1": 1, "yes": 1, "y": 1,
    "0": 0, "no": 0, "n": 0
})

df["HasCrCard"] = df["HasCrCard"].astype(int)

print(df["HasCrCard"].value_counts())
```

```python
df["IsActiveMember"].value_counts()
```

```python

df["IsActiveMember"] = df["IsActiveMember"].astype(str).str.strip().str.lower()

df["IsActiveMember"] = df["IsActiveMember"].replace({
    "1": 1, "yes": 1,
    "0": 0, "no": 0
})

df["IsActiveMember"] = df["IsActiveMember"].astype(int)

print(df["IsActiveMember"].value_counts())
```

```python
print("Missing:", df["EstimatedSalary"].isnull().sum())
print(df["EstimatedSalary"].describe())
```

```python
median_salary = df["EstimatedSalary"].median()
df["EstimatedSalary"] = df["EstimatedSalary"].fillna(median_salary)

print("Median used:", median_salary)
print("Remaining missing:", df["EstimatedSalary"].isnull().sum())
```

```python
# Flag impossible negative salaries as missing
df.loc[df["EstimatedSalary"] < 0, "EstimatedSalary"] = np.nan

print("Missing after flagging negatives:", df["EstimatedSalary"].isnull().sum())

# Re-fill with median (recalculated on the now-cleaner data)
median_salary = df["EstimatedSalary"].median()
df["EstimatedSalary"] = df["EstimatedSalary"].fillna(median_salary)

print("Median used:", median_salary)
print("Remaining missing:", df["EstimatedSalary"].isnull().sum())
print(df["EstimatedSalary"].describe())
```

```python
df["Branch"].value_counts()
```

```python
df["Branch"] = df["Branch"].str.strip().str.lower()
df["Branch"] = df["Branch"].str.title()

print(df["Branch"].value_counts())
```

```python
df["JoinDate"].head(20)
```

```python
df["JoinDate"] = pd.to_datetime(df["JoinDate"], format="mixed", dayfirst=True)

print(df["JoinDate"].dtype)
df["JoinDate"].head(20)
```

```python
print("Missing:", df["Exited"].isnull().sum())
df["Exited"].value_counts()
```

```python
df.info()
print("\nTotal missing values across all columns:")
print(df.isnull().sum().sum())
print("\nDuplicate rows remaining:", df.duplicated().sum())
print("\nRow count:", len(df))
print("\nFinal dtypes:")
print(df.dtypes)
```
